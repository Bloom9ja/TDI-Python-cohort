```python
# Personal Information (Example Data - Safe to Share)
first_name = "Alex"
last_name = "Johnson"
age = 28
email = "alex.johnson@example.com"
city = "Toronto"
country = "Canada"
```

```python
# Personal Profile
print("Name: " + first_name + " " + last_name)
print("Age: " + str(age))
print("Email: " + email)
print("Location: " + city + ", " + country)
```

```python
# 1. How old will Alex be in 10 years?
future_age = age + 10
print("In 10 years: " + str(future_age) + " years old")

# 2. How old was Alex 5 years ago?
past_age = age - 5
print("5 years ago: " + str(past_age) + " years old")

# 3. How many days has Alex been alive? (rough estimate)
days_alive = age * 365
print("Days alive: ~" + str(days_alive) + " days")

# 4. How many years until retirement at 65?
years_to_retirement = 65 - age
print("Years to retirement: " + str(years_to_retirement) + " years")
```

```python
# Check the data type
print(type(first_name))
print(type(age))
print(type(email))
print(type(days_alive))
```
